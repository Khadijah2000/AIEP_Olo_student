### Task 2.1 — Return rate by every category, on one figure

Build a figure with one panel per categorical column — `channel`, `customer_tier` and
`city_clean` — each showing the return rate per category as a bar chart.

Draw the overall return rate as a horizontal reference line on every panel. Without it a reader
cannot tell whether a bar is high or the whole dataset is high, and that reference line is the
difference between a chart that informs and a chart that decorates.

<div dir="rtl" align="right">

### المهمة ٢٫١ — نسبة الإرجاع لكل فئة في شكل واحد

ابنِ شكلًا فيه لوحة لكل عمود فئوي — `channel` و`customer_tier` و`city_clean` — تُظهر كل لوحة نسبة
الإرجاع لكل فئة كرسم أعمدة.

وارسم نسبة الإرجاع الإجمالية كخط مرجعي أفقي في كل لوحة. فبدونه لا يستطيع القارئ تمييز ما إذا كان
العمود مرتفعًا أم أن المجموعة كلها مرتفعة، وهذا الخط المرجعي هو الفرق بين رسم يُعلم ورسم يزيّن.

</div>

```python
# ────────────────────────────────────────────────────────────────────
# 1) Make one row of subplots, one column of the figure per categorical column.
# 2) For each, group by that column and take the mean of `returned` — that is the rate.
# 3) Plot it as bars, then add the overall mean as a horizontal line on the same axes.
# 4) Label the y axis as a share and give every panel a title, or the figure is unreadable.
# Search: "matplotlib subplots axhline bar"
# https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.axhline.html
#
# ١) أنشئ صفًّا واحدًا من اللوحات الفرعية، لوحة لكل عمود فئوي.
# ٢) جمّع لكل منها حسب ذلك العمود وخذ متوسّط `returned` — فهو النسبة.
# ٣) ارسمها أعمدة، ثم أضف المتوسّط الإجمالي كخط أفقي على المحاور نفسها.
# ٤) سمِّ المحور الرأسي كحصّة وأعطِ كل لوحة عنوانًا، وإلا تعذّرت قراءة الشكل.
# ابحث عن: "matplotlib subplots axhline bar"
# ────────────────────────────────────────────────────────────────────

CATEGORICAL = ["channel", "customer_tier", "city_clean"]
overall = raw["returned"].mean()
# TODO: Draw one bar panel per categorical column showing the return rate of each category, with the overall rate as a horizontal reference line on each panel.
# مهمة: ارسم لوحة أعمدة لكل عمود فئوي تُظهر نسبة الإرجاع لكل فئة، مع نسبة الإرجاع الإجمالية كخط مرجعي أفقي على كل لوحة.
# حسبت نسبة الإرجاع لكل فئة عشان أشوف إذا فيه فئات ترجع طلبات أكثر من غيرها.
# استخدمت القناة ونوع العميل والمدينة للمقارنة بين المجموعات.
# أضفت المعدل العام عشان أقارن كل فئة بالنسبة الإجمالية للإرجاع.

rate_tables = {}

for col in CATEGORICAL:
    # حسبت متوسط returned لكل فئة، وهو يمثل نسبة الإرجاع.
    rate_tables[col] = raw.groupby(col)["returned"].mean()

# أنشأت ثلاث لوحات للرسم، لوحة لكل عمود فئوي.
fig, axes = plt.subplots(
    1,
    len(CATEGORICAL),
    figsize=(15, 4),
    sharey=True
)

for ax, col in zip(axes, CATEGORICAL):
    rates = rate_tables[col]

    # رسمت نسبة الإرجاع لكل فئة على شكل أعمدة.
    ax.bar(rates.index.astype(str), rates.values)

    # أضفت خط يوضح نسبة الإرجاع العامة لجميع الطلبات.
    ax.axhline(
        overall,
        linestyle="--",
        linewidth=2,
        label=f"Overall: {overall:.1%}"
    )

    # أضفت عناوين وتسميات عشان يكون الرسم واضح.
    ax.set_title(f"Return rate by {col}")
    ax.set_xlabel(col)
    ax.set_ylabel("Return rate")

    # لفيت أسماء الفئات عشان ما تتداخل مع بعض.
    ax.tick_params(axis="x", rotation=45)

    # حددت نطاق المحور الرأسي عشان تظهر الأعمدة بشكل مناسب.
    ax.set_ylim(0, max(rates.max(), overall) + 0.1)

    # أظهرت وسيلة الإيضاح الخاصة بالمعدل العام.
    ax.legend()

plt.tight_layout()
fig.savefig(CHART_DIR / "return_rate_by_category.png", bbox_inches="tight")
plt.show()

# طبعت نتائج كل عمود عشان أقدر أقارن نسب الإرجاع بين الفئات.
for col, rates in rate_tables.items():
    print(f"--- {col} ---")
    print(rates.to_string(float_format="{:.1%}".format), "\n")
```

### المهمة ٢٫٢ — هل الفرق كبير بما يكفي للتصديق؟

فئتك الأفضل في المهمة ٢٫١ لها نسبة إرجاع أعلى من الأسوأ. وقبل أن تصدّق ذلك، تحقّق من عدد الطلبات

```python
# ────────────────────────────────────────────────────────────────────
# 1) `agg` can compute several things at once — ask it for the mean and the size.
# 2) Name the outputs so the columns read as "return_rate" and "orders".
# 3) Sort by rate so the extremes are at the ends, and print all three tables.
# Search: "pandas groupby agg named aggregation"
# https://pandas.pydata.org/docs/user_guide/groupby.html#named-aggregation
#
# ١) يستطيع `agg` حساب أشياء عدة دفعة واحدة — اطلب منه المتوسّط والحجم.
# ٢) سمِّ المخرجات بحيث تُقرأ الأعمدة «return_rate» و«orders».
# ٣) رتّب حسب النسبة ليكون الطرفان في النهايتين، واطبع الجداول الثلاثة.
# ابحث عن: "pandas groupby agg named aggregation"
# ────────────────────────────────────────────────────────────────────

# TODO: For each categorical column build a table of return rate and order count, sorted by rate, and print it.
# مهمة: ابنِ لكل عمود فئوي جدولًا بنسبة الإرجاع وعدد الطلبات، مرتّبًا حسب النسبة، واطبعه.
# أنشأت قاموس عشان أحفظ جدول كل عمود فئوي.
group_tables = {}

# مررت على كل عمود فئوي وحسبت نسبة الإرجاع وعدد الطلبات لكل فئة.
for col in CATEGORICAL:
    group_tables[col] = (
        raw.groupby(col)["returned"]
        .agg(
            return_rate="mean",
            orders="size"
        )
        .sort_values("return_rate")
    )

# طبعت الجداول الثلاثة بعد ترتيبها حسب نسبة الإرجاع.
for col, table in group_tables.items():
    print(f"--- {col} ---")
    print(
        table.to_string(
            formatters={"return_rate": "{:.1%}".format}
        ),
        "\n"
    )
```

**الصلة بمشروعك:** حين تصل إلى هندسة الخصائص في مشروعك، تكون التفاعلات التي تجدها بهذه الطريقة هي

```python
# ────────────────────────────────────────────────────────────────────
# There is no single right answer. Pick two columns, cut any continuous one into a
# few buckets, then group by both and take the mean of `returned`. A pivot table or
# a heatmap reads better than a long list. Print the group sizes too — a two-way
# split makes small groups fast, and a rate over 20 rows is not a finding.
#
# لا توجد إجابة واحدة صحيحة. اختر عمودين، وقسّم أي عمود متّصل إلى بضع سلال، ثم
# جمّع حسب الاثنين وخذ متوسّط `returned`. والجدول المحوري أو الخريطة الحرارية
# أسهل قراءة من قائمة طويلة. واطبع أحجام المجموعات أيضًا — فالتقسيم الثنائي
# يصغّر المجموعات بسرعة، والنسبة على عشرين صفًّا ليست نتيجة.
# ────────────────────────────────────────────────────────────────────

# TODO: Investigate one two-column question and show the result as a pivot table or chart.
# مهمة: حقّق في سؤال من عمودين وأظهر النتيجة كجدول محوري أو رسم.
# اخترت سؤال يشوف العلاقة بين نسبة الخصم وقناة الطلب.
# أبي أعرف هل الخصومات مرتبطة بالإرجاع بشكل مختلف حسب القناة.

# قسمت نسبة الخصم إلى ثلاث فئات عشان يكون التحليل أسهل في القراءة.
raw["discount_level"] = pd.cut(
    raw["discount_pct"],
    bins=[-np.inf, 0.10, 0.20, np.inf],
    labels=["low", "medium", "high"]
)

# جمعت البيانات حسب قناة الطلب ومستوى الخصم.
# أخذت متوسط returned لأنه يمثل نسبة الإرجاع.
two_way = (
    raw.groupby(["channel", "discount_level"], observed=True)["returned"]
    .agg(
        return_rate="mean",
        orders="size"
    )
    .reset_index()
)

# أنشأت جدول محوري عشان أقارن نسب الإرجاع بين القنوات ومستويات الخصم.
pivot = two_way.pivot(
    index="discount_level",
    columns="channel",
    values="return_rate"
)

print("Return rate by discount level and channel:")
print(
    pivot.to_string(
        float_format="{:.1%}".format
    )
)

print("\nGroup sizes:")
print(
    two_way.pivot(
        index="discount_level",
        columns="channel",
        values="orders"
    )
)

# رسمت النتائج عشان تكون المقارنة بين القنوات ومستويات الخصم أوضح.
pivot.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Return Rate by Discount Level and Channel")
plt.xlabel("Discount level")
plt.ylabel("Return rate")
plt.xticks(rotation=0)
plt.legend(title="Channel")

plt.tight_layout()
fig.savefig(CHART_DIR / "اسم_الرسم.png", bbox_inches="tight")
plt.show()
```
